# Создаем список ссылок на файлы .txt

```python
# URL сайта, где находятся файлы
url = 'https://www.ngdc.noaa.gov/stp/space-weather/swpc-products/annual_reports/daily_solar_indices_summaries/daily_solar_data/'  # Замените на ваш URL

# Получаем HTML-контент страницы
response = requests.get(url)
soup = BeautifulSoup(response.text, 'html.parser')

# Находим все ссылки на файлы .txt
links = []
for a in soup.find_all('a', href=True):
    if a['href'].endswith('.txt'):
        links.append(a['href'])

# Если ссылки относительные, сделаем их абсолютными
base_url = url.rsplit('/', 1)[0]  # Получаем базовый URL
full_links = [link if link.startswith('http') else f"{base_url}/{link}" for link in links]
full_links.reverse()
```

# Сохраняем файлы .txt в csv

```python
# Список для хранения данных
dataframes = []
k = 0
# Загружаем каждый файл и добавляем его в список DataFrame
for link in full_links:
    k = k+1
    try:
        print('Итерация:', k, '\nСсылка:', link) 
        print()
        # Скачиваем файл
        response = requests.get(link)
        data = response.text

        # Используем StringIO для обработки строк как файла
        data_io = StringIO(data)
        lines = data_io.readlines()
        
        # Ищем индекс строки, где начинаются данные (первая строка с данными)
        start_index = None
        for i, line in enumerate(lines):
            if line.startswith("2") or line.startswith("1") or line.startswith("0") or line.startswith("3"):
                start_index = i
                break
        
        # Если нашли индекс, создаем DataFrame
        if start_index is not None:
            # Читаем данные в DataFrame, пропуская строки до start_index
            df = pd.read_csv(StringIO(''.join(lines[start_index:])),delim_whitespace=True) 
            
            # Сохраняем DataFrame в CSV
            df.to_csv(f'daily_solar_data_{k}.csv', index=False) 
        
            print(f"Данные успешно сохранены в daily_solar_data_{k}.csv!\n")
        else:
            print("Данные не найдены.")
    except Exception as e:
        print(f"Ошибка при обработке {link}: {e}")
```

```python
import pandas as pd
import seaborn as sns
from pandasql import sqldf
```

```python
query = """
select 
     cast(case when length(cast(month as varchar)) = 1 and length(cast(day as varchar)) = 1  then CONCAT(cast(year as varchar), '-0', cast(month as varchar),  '-0',  cast(day as varchar))
               when length(cast(month as varchar)) = 1 and length(cast(day as varchar)) = 2  then CONCAT(cast(year as varchar), '-0', cast(month as varchar),  '-',  cast(day as varchar))
               when length(cast(month as varchar)) = 2 and length(cast(day as varchar)) = 1  then CONCAT(cast(year as varchar), '-', cast(month as varchar),  '-0',  cast(day as varchar))
               when length(cast(month as varchar)) = 2 and length(cast(day as varchar)) = 2  then CONCAT(cast(year as varchar), '-', cast(month as varchar),  '-',  cast(day as varchar))
      end as varchar) as calc_date
 , radio_flux
 , sunspot_number
 , sunspot_area
 , new_region
 , stanford_solar_mean_field
 , flares_x_ray_flux_bkgd
 , flares_x_ray_c
 , flares_x_ray_m
 , flares_x_ray_x
 , flares_optical_s
 , flares_optical_1
 , flares_optical_2
 , flares_optical_3                       
from df_no_duplicates
where length(cast(year as varchar)) = 4
;
"""
df_fin = sqldf(query)
```
